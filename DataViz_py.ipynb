```python
import matplotlib.pyplot as plt
from random import choice
class RandomWalk:
  # A class to generate random walks.
  def __init__(self, num_points=5000):
    # Initialize attributes of a walk.
    self.num_points = num_points
    # All walks start at (0, 0).
    self.x_values = [0]
    self.y_values = [0]
  def get_step(self):
    # Calculate all steps in the walk.
    direction = choice([1, -1])
    distance = choice([0, 1, 2, 3, 4])
    return direction * distance
  def fill_walk(self):
    # Determine the direction and distance for each step.
    while len(self.x_values) < self.num_points:
      x_step = self.get_step()
      y_step = self.get_step()
    # Reject moves that go nowhere.
      if x_step == 0 and y_step == 0:
        continue
    # Calculate the new position.
      x = self.x_values[-1] + x_step
      y = self.y_values[-1] + y_step

      self.x_values.append(x)
      self.y_values.append(y)

# Plot the random walk
rw = RandomWalk()
rw.fill_walk()
fig, ax = plt.subplots()
point_numbers = range(rw.num_points)
ax.plot(rw.x_values, rw.y_values, c='red', linewidth=3)
# Emphasize the first and last points.
ax.plot(0, 0, c='green', label='First Walk', marker='o', linewidth=3)
ax.plot(rw.x_values[-1], rw.y_values[-1], c='black', marker='o', label='Last Walk', linewidth=3)
ax.legend()
ax.get_xaxis().set_visible(False)
ax.get_yaxis().set_visible(False)
plt.show()
```

```python
! python -m pip install --user plotly
from random import randint
class Die:
  def __init__(self, num_sides=6):
    self.num_sides = num_sides
  def roll(self):
    return randint(1, self.num_sides)

# Roll a die.
die_1 = Die()
die_2 = Die()
results = []
for roll_num in range(1000):
  result = die_1.roll() * die_2.roll()
  results.append(result)
# Analyze the results.
frequencies = []
max_result = die_1.num_sides * die_2.num_sides
for value in range(1, max_result + 1):
  frequency = results.count(value)
  frequencies.append(frequency)
print(frequencies)
```

```python
import matplotlib.pyplot as plt
x_values = list(range(1, max_result + 1))
y_values = frequencies
plt.bar(x_values, y_values, color='red', edgecolor='black')
plt.title("Results of Randomly Rolling Two D6 1000 Times", fontsize=20)
plt.xlabel("Die Face", fontsize=10)
plt.ylabel("Frequency", fontsize=10)
plt.show()
```
